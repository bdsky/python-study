### numpy基础
- numpy是专门用于数值计算的数组结构，比列表类型封装了更多数值计算相关的操作
- numpy数组要求元素类型统一，与list元素类型可以有多种不同
- 对numpy变量进行+，-，*，/，** 是作用在内部每个元素中，与列表不同
- 与pandas类似，numpy可以进行条件运算并进行布尔索引

```python
import numpy as np
#列表类型
a = [1,3,4]
print(f'{a=}, {type(a)=}')
print(f'{a * 2 = }')
print('-'*40)
#numpy类型
b = np.array([1,2,3,4])
print(f'{b = }')
print(f'{type(b) = }')
print(f'{b * 2 = }')
print('-'*40)
#条件运算, 布尔索引
print(f'{b > 2 = }, {type(b > 2) = }')
print(f'{b[b > 2]=}, {type(b[b > 2]) = }')
##使用布尔索引修改值
#print(b[b > 2] = 0)  在python中赋值语句不是表达式，没有返回值，也不是返回None
print(f'{b = }, {type(b) = }')
```
